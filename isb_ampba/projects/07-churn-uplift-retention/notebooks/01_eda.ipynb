# 01 - EDA: transactions, cleaning, snapshots
Thin notebook: all logic lives in `src/churnx`.

```python
import sys; sys.path.insert(0, '../src')
from churnx.config import load_config
cfg = load_config()
from churnx.data import load_transactions
from churnx.snapshot import clean_sales, build_all_snapshots
tx = load_transactions(cfg)
sales, returns = clean_sales(tx)
print(len(sales), sales.customer_id.nunique(), len(returns))
```

```python
snaps = build_all_snapshots(sales, returns, cfg)
snaps.groupby(['asof','split']).churn.agg(['size','mean'])
```
